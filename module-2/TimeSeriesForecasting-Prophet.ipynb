# Time Series Forecasting with Prophet

```python
import pandas as pd 
from prophet import Prophet
```

```python
df = pd.read_csv("manning.csv", usecols=["ds", "y"])
df.head()
```

```python
df.tail()
```

```python
m = Prophet() # Create an instance of the prophet model
m.fit(df) #fit the model to the dataset
```

```python
future = m.make_future_dataframe(periods = 365)
```

```python
future.tail(3)
```

```python
forecast = m.predict(future)
forecast.tail()
```

```python
forecast[["ds", "yhat", "yhat_lower", "yhat_upper"]].tail()
```

```python
m.plot(forecast)
```

```python
m.plot_components(forecast)
```
